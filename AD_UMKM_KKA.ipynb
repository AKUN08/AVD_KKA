```python
df = pd.read_csv("umkm_kka.csv")

print("LIMA DATA PERTAMA:")
display(df.head())
```

```python
print("Jumlah baris dan kolom:")
print(df.shape)

print("\nJumlah baris:", df.shape[0])
print("Jumlah kolom:", df.shape[1])
```

```python
df["timestamp_transaksi"] = pd.to_datetime(
    df["timestamp_transaksi"],
    errors="coerce"
)

print(df["timestamp_transaksi"].dtype)
```

```python
df.to_csv("dataset_umkm_bersih.csv", index=False)

print("Dataset bersih berhasil disimpan!")
```

```python
print("STATISTIK DESKRIPTIF:")
display(df.describe())
```

```python
total_penjualan = df["nilai_transaksi_netto_idr"].sum()

print("Total Penjualan:")
print(f"Rp {total_penjualan:,.0f}")
```

```python
rata_rata = df["nilai_transaksi_netto_idr"].mean()

print("Rata-rata Transaksi:")
print(f"Rp {rata_rata:,.0f}")
```

```python
median = df["nilai_transaksi_netto_idr"].median()

print("Median Transaksi:")
print(f"Rp {median:,.0f}")
```

```python
std_deviasi = df["nilai_transaksi_netto_idr"].std()

print("Standar Deviasi:")
print(f"Rp {std_deviasi:,.2f}")
```

```python
kategori_terlaris = (
    df.groupby("kategori_produk")["nilai_transaksi_netto_idr"]
    .sum()
    .sort_values(ascending=False)
)

display(kategori_terlaris)

print("\nKategori terlaris:")
print(kategori_terlaris.idxmax())
```

```python
wilayah_tertinggi = (
    df.groupby("provinsi_pembeli")["nilai_transaksi_netto_idr"]
    .sum()
    .sort_values(ascending=False)
)

display(wilayah_tertinggi)

print("\nWilayah penjualan tertinggi:")
print(wilayah_tertinggi.idxmax())
```

```python
rating = df["rating_produk"].mean()

print("Rata-rata Rating:")
print(round(rating, 2))
```

```python
data = df["nilai_transaksi_netto_idr"]

Q1 = data.quantile(0.25)
Q2 = data.quantile(0.50)
Q3 = data.quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
```

VISUALISASI DATA

```python
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import FuncFormatter

sns.set_style("whitegrid")

# Format sumbu Rupiah (500 jt, 1.2 M)
def format_rupiah(x, pos):
    if x >= 1e9:
        return f"{x/1e9:.1f} M"
    elif x >= 1e6:
        return f"{x/1e6:.0f} jt"
    elif x >= 1e3:
        return f"{x/1e3:.0f} rb"
    return f"{x:.0f}"

fmt = FuncFormatter(format_rupiah)

print("Library visualisasi berhasil diimpor!")
```

```python
df["bulan"] = df["timestamp_transaksi"].dt.to_period("M")

tren_bulanan = df.groupby("bulan")["nilai_transaksi_netto_idr"].sum()

print("Tren penjualan bulanan:")
display(tren_bulanan.head())
```

Grafik 1: Bar Chart, Total Penjualan per Kategori Produk

Grafik ini membandingkan total penjualan (nilai netto) dari 7 kategori produk UMKM.

```python
plt.figure(figsize=(9, 5))
sns.barplot(x=kategori_terlaris.values, y=kategori_terlaris.index,
            hue=kategori_terlaris.index, palette="Blues_r", legend=False)
plt.title("Total Penjualan per Kategori Produk")
plt.xlabel("Total Penjualan Netto (Rp)")
plt.ylabel("Kategori Produk")
plt.gca().xaxis.set_major_formatter(fmt)
plt.tight_layout()
plt.show()

print("Kategori tertinggi:", kategori_terlaris.idxmax())
print("Kategori terendah :", kategori_terlaris.idxmin())
```

Grafik 2: Line Chart, Tren Penjualan Bulanan

Grafik ini menunjukkan naik-turunnya total penjualan dari bulan ke bulan (Februari 2024 sampai Oktober 2026).

```python
plt.figure(figsize=(11, 5))
plt.plot(tren_bulanan.index.astype(str), tren_bulanan.values,
         marker="o", color="crimson")
plt.title("Tren Penjualan Bulanan")
plt.xlabel("Bulan")
plt.ylabel("Total Penjualan Netto (Rp)")
plt.gca().yaxis.set_major_formatter(fmt)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("Bulan tertinggi:", tren_bulanan.idxmax())
print("Bulan terendah :", tren_bulanan.idxmin())
```

Grafik 3: Scatter Plot, Nilai Transaksi Bruto vs Ongkos Kirim

Grafik ini melihat apakah harga belanjaan berhubungan dengan ongkos kirimnya. Warna titik menunjukkan wilayah pengiriman.

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x="nilai_transaksi_bruto_idr", y="ongkos_kirim_riil_idr",
                hue="wilayah_pengiriman", alpha=0.6)
plt.title("Hubungan Nilai Transaksi Bruto dengan Ongkos Kirim")
plt.xlabel("Nilai Transaksi Bruto (Rp)")
plt.ylabel("Ongkos Kirim Riil (Rp)")
plt.gca().xaxis.set_major_formatter(fmt)
plt.gca().yaxis.set_major_formatter(fmt)
plt.legend(title="Wilayah Pengiriman", fontsize=8)
plt.tight_layout()
plt.show()

korelasi = df["nilai_transaksi_bruto_idr"].corr(df["ongkos_kirim_riil_idr"])
print("Korelasi nilai bruto vs ongkir:", round(korelasi, 3))
```

Grafik 4: Histogram, Distribusi Rating Produk

Grafik ini menunjukkan berapa banyak transaksi untuk setiap rating (1 sampai 5).

```python
plt.figure(figsize=(7, 5))
sns.histplot(df["rating_produk"], bins=5, color="seagreen")
plt.title("Distribusi Rating Produk")
plt.xlabel("Rating")
plt.ylabel("Frekuensi")
plt.show()

print("Rata-rata rating:", round(df["rating_produk"].mean(), 2))
print(df["rating_produk"].value_counts().sort_index())
```

Grafik 5: Boxplot, Sebaran Nilai Transaksi per Kategori

Grafik ini menunjukkan sebaran nilai transaksi tiap kategori. Kotak adalah 50% data di tengah, garis tengah adalah median, dan titik di luar kotak adalah outlier (transaksi yang nilainya jauh lebih besar dari biasanya).

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="kategori_produk", y="nilai_transaksi_netto_idr")
plt.title("Sebaran Nilai Transaksi per Kategori Produk")
plt.xlabel("Kategori Produk")
plt.ylabel("Nilai Transaksi Netto (Rp)")
plt.gca().yaxis.set_major_formatter(fmt)
plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()
```

Grafik 6: Bar Chart, 10 Provinsi Pembeli dengan Penjualan Tertinggi

Grafik ini menunjukkan dari provinsi mana saja penjualan paling banyak datang.

```python
top10_wilayah = wilayah_tertinggi.head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top10_wilayah.values, y=top10_wilayah.index,
            hue=top10_wilayah.index, palette="viridis", legend=False)
plt.title("10 Provinsi Pembeli dengan Penjualan Tertinggi")
plt.xlabel("Total Penjualan Netto (Rp)")
plt.ylabel("Provinsi Pembeli")
plt.gca().xaxis.set_major_formatter(fmt)
plt.tight_layout()
plt.show()

print("Wilayah penjualan tertinggi:", wilayah_tertinggi.idxmax())
print(f"Total: Rp {wilayah_tertinggi.max():,.0f}")

# Kontribusi 6 provinsi teratas terhadap total penjualan
kontribusi = wilayah_tertinggi.head(6).sum() / wilayah_tertinggi.sum() * 100
print(f"Kontribusi 6 provinsi teratas: {kontribusi:.1f}% dari total penjualan")
```

```python
# Inisialisasi subplots dengan 3 baris dan 2 kolom (total 6 grafik)
fig, axes = plt.subplots(3, 2, figsize=(15, 18))

total_kategori = (
    df.groupby('kategori_produk')['nilai_transaksi_bruto_idr'].sum().sort_values()
)
sns.barplot(x=total_kategori.values, y=total_kategori.index, ax=axes[0, 0])
axes[0, 0].set_title('Total Penjualan per Kategori')
axes[0, 0].set_xlabel('Total Nilai Transaksi Bruto (IDR)')
axes[0, 0].set_ylabel('Kategori Produk')


df['timestamp_transaksi'] = pd.to_datetime(df['timestamp_transaksi'])
df['bulan_tahun'] = df['timestamp_transaksi'].dt.to_period('M')
tren_bulanan = df.groupby('bulan_tahun')['nilai_transaksi_bruto_idr'].sum()

axes[0, 1].plot(
    tren_bulanan.index.astype(str), tren_bulanan.values, marker='o'
)
axes[0, 1].set_title('Tren Penjualan Bulanan')
axes[0, 1].set_xlabel('Bulan')
axes[0, 1].set_ylabel('Total Penjualan (IDR)')
axes[0, 1].tick_params(axis='x', rotation=45)


sns.histplot(df['rating_produk'], bins=5, ax=axes[1, 0], discrete=True)
axes[1, 0].set_title('Distribusi Rating Pelanggan')
axes[1, 0].set_xlabel('Rating Produk')
axes[1, 0].set_ylabel('Frekuensi')


sns.boxplot(
    data=df,
    x='kategori_produk',
    y='nilai_transaksi_bruto_idr',
    ax=axes[1, 1],
)
axes[1, 1].set_title('Sebaran Jumlah per Kategori')
axes[1, 1].set_xlabel('Kategori Produk')
axes[1, 1].set_ylabel('Nilai Transaksi Bruto (IDR)')
axes[1, 1].tick_params(axis='x', rotation=45)


sns.scatterplot(
    data=df,
    x='nilai_transaksi_bruto_idr',
    y='ongkos_kirim_riil_idr',
    hue='wilayah_pengiriman',
    alpha=0.6,
    ax=axes[2, 0],
)
axes[2, 0].set_title('Hubungan Nilai Transaksi Bruto dengan Ongkos Kirim')
axes[2, 0].set_xlabel('Nilai Transaksi Bruto (Rp)')
axes[2, 0].set_ylabel('Ongkos Kirim Riil (Rp)')
axes[2, 0].xaxis.set_major_formatter(fmt)
axes[2, 0].yaxis.set_major_formatter(fmt)
axes[2, 0].legend(title='Wilayah Pengiriman', fontsize=7)


wilayah_tertinggi = (
    df.groupby('provinsi_pembeli')['nilai_transaksi_netto_idr']
    .sum()
    .sort_values(ascending=False)
)
top10_wilayah = wilayah_tertinggi.head(10)

sns.barplot(
    x=top10_wilayah.values, y=top10_wilayah.index, palette='viridis', ax=axes[2, 1]
)
axes[2, 1].set_title('10 Provinsi Pembeli dengan Penjualan Tertinggi')
axes[2, 1].set_xlabel('Total Penjualan Netto (Rp)')
axes[2, 1].set_ylabel('Provinsi Pembeli')
axes[2, 1].xaxis.set_major_formatter(fmt)
axes[2, 1].tick_params(axis='y', labelsize=8)


plt.tight_layout()
plt.show()
```

```python
total_penjualan = df['nilai_transaksi_netto_idr'].sum()
rata_transaksi = df['nilai_transaksi_netto_idr'].mean()
kategori_terlaris = (
    df.groupby('kategori_produk')['nilai_transaksi_netto_idr']
    .sum()
    .idxmax()
)
wilayah_teratas = (
    df.groupby('provinsi_pembeli')['nilai_transaksi_netto_idr'].sum().idxmax()
)
rating_rerata = df['rating_produk'].mean()


total_penjualan_str = f'Rp {total_penjualan:,.0f}'.replace(',', '.')
rata_transaksi_str = f'Rp {rata_transaksi:,.0f}'.replace(',', '.')


print('=== LAPORAN ANALISIS DATA PENJUALAN NUSANTARA MART ===\n')

print('1. Ringkasan Statistik')
print(f'   Total Penjualan     : {total_penjualan_str}')
print(f'   Rata-rata Transaksi : {rata_transaksi_str}')
print(f'   Kategori Terlaris   : {kategori_terlaris}')
print(f'   Wilayah Teratas     : {wilayah_teratas}')
print(f'   Rating Rata-rata    : {rating_rerata:.2f} / 5\n')

print('2. Temuan Utama (Insight)')
print('   - Penjualan didominasi oleh kategori Fashion Muslim & Batik serta Kriya & Kerajinan Tangan.')
print('   - Wilayah DI Yogyakarta, Jawa Timur, dan DKI Jakarta menyumbang volume transaksi netto tertinggi.')
print('   - Mayoritas pengiriman tepat waktu dengan tingkat kepuasan pelanggan yang solid.')
print('\n3. Potensi Outlier')
print('   - Terdapat anomali nilai transaksi bruto yang sangat tinggi pada beberapa transaksi korporat/grosir.')
print('\n4. Rekomendasi Bisnis')
print('   - Mengoptimalkan subsidi ongkir dan memperluas ekspansi logistik di'' luar Pulau Jawa untuk menekan risiko churn.')
```
